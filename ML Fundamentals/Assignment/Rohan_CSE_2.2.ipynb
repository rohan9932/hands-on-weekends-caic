```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

# Inspect Data

```python
df=load_wine(as_frame=True)
df=df.frame
df.head(10)
```

```python
df.info()
```

```python
df.columns
```

```python
df['target'].value_counts()
```

# Train Data

```python
x_cls=df.drop(['target'], axis=1)
y_cls=df['target']
x_train_cls,x_test_cls,y_train_cls,y_test_cls=train_test_split(x_cls,y_cls,test_size=0.2, random_state=42)
```

# Model Build

```python
log=make_pipeline(StandardScaler(),LogisticRegression()) # for scaling different colums automatically
log.fit(x_train_cls, y_train_cls)
y_pred=log.predict(x_test_cls)

print(f"First 10 predictions: {y_pred[:10]}")
print(f"First 10 actual values: {y_test_cls[:10].values}")
```

# Evaluation

```python
accuracy=accuracy_score(y_test_cls,y_pred)
accuracy
```

```python
precision=precision_score(y_test_cls,y_pred, average='macro')
precision
```

```python
recall=recall_score(y_test_cls,y_pred, average='macro')
recall
```
